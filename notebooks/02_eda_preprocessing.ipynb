```python
import sys, os
from pathlib import Path

# Make the project root importable so `from src... import ...` works
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.append(str(PROJECT_ROOT))

from dotenv import load_dotenv
load_dotenv(PROJECT_ROOT / ".env")

print("Project root:", PROJECT_ROOT)
```

```python
from src.parse_data import load_all_raw_reviews

raw_df = load_all_raw_reviews()
print(f"Loaded {len(raw_df)} unique reviews")
raw_df.head()
```

## Explore the raw data

```python
import matplotlib.pyplot as plt

raw_df["star_rating"].value_counts().sort_index().plot(kind="bar", title="Star rating distribution")
plt.xlabel("Stars"); plt.ylabel("Count")
plt.show()

print("Verified purchase rate:", raw_df["verified_purchase"].mean())
print("Reviews with images:", raw_df["has_images"].sum())
```

## Clean, filter, and label

```python
from src.preprocess import build_dataset, balance_classes

df = build_dataset(raw_df)
df["sentiment"].value_counts()
```

```python
df_balanced = balance_classes(df)
print("Class distribution after balancing:")
print(df_balanced["sentiment"].value_counts())
df_balanced.head()
```

## Save processed dataset

```python
from src.config import PROCESSED_DATA_DIR

out_path = PROCESSED_DATA_DIR / "dataset.csv"
df_balanced.to_csv(out_path, index=False)
print(f"Saved {len(df_balanced)} rows to {out_path}")
```
