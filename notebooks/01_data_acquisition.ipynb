```python
import sys, os
from pathlib import Path

# Make the project root importable so `from src... import ...` works
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.append(str(PROJECT_ROOT))

from dotenv import load_dotenv
load_dotenv(PROJECT_ROOT / ".env")

print("Project root:", PROJECT_ROOT)
```

```python
from src.fetch_data import fetch_all_pages

# Increase NUM_PAGES for a bigger dataset. You'll want 500-1000+ reviews
# total for the transformer model in notebook 04 to learn anything useful.
NUM_PAGES = 30
COUNTRY = "IN"

saved_paths = fetch_all_pages(num_pages=NUM_PAGES, country=COUNTRY)
print(f"Saved {len(saved_paths)} pages.")
```

```python
# Quick sanity check: load one page back and peek at it
import json
with open(saved_paths[0]) as f:
    sample = json.load(f)

print("Total reviews on product:", sample["data"].get("total_reviews"))
print("Rating distribution:", sample["data"].get("rating_distribution"))
print("Reviews in this page:", len(sample["data"].get("reviews", [])))
```
